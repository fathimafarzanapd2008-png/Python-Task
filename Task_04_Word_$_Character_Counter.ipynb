```python
# Task 4: Word & Character Counter

print("=" * 50)
print("          WORD & CHARACTER COUNTER          ")
print("=" * 50)

# Input text from user
text = input("Enter a paragraph or sentence: ")

# Calculations
char_count_with_spaces = len(text)
char_count_no_spaces = len(text.replace(" ", ""))
words = text.split()
word_count = len(words)
sentence_count = text.count('.') + text.count('!') + text.count('?')

# Output summary
print("\n" + "=" * 50)
print("                 ANALYSIS RESULT                ")
print("=" * 50)
print(f"Total Characters (with spaces)   : {char_count_with_spaces}")
print(f"Total Characters (without spaces): {char_count_no_spaces}")
print(f"Total Words                      : {word_count}")
print(f"Estimated Sentences              : {sentence_count}")
print("=" * 50)
```
